```python
import pandas as pd
import matplotlib.pyplot as plt
df = pd.read_csv("Sheet1.csv")
```

```python
df.head()
```

## Data Exploration
Taking a look at the data to see the patterns.

```python
fig, ax = plt.subplots(figsize=(10,5))
df.groupby(['shop_id']).mean()['order_amount'].plot(ax=ax)
```

```python
fig, ax = plt.subplots(figsize=(10,4))
df.groupby(['user_id']).median()['total_items'].plot(ax=ax)
```

Certain users are ordering many more items, characteristically looking like a bulk order

```python
fig, ax = plt.subplots(figsize=(10,5))
df.groupby(['shop_id']).count()['total_items'].plot(ax=ax)
```

Shops get around 50 orders a month with no large outliers

## Quantifiers

```python
print("Median order amount:", df['order_amount'].median())
print("Median order amount:", df['order_amount'].mean())
print("Mode order amount:", df['order_amount'].mode().iat[0])
```

```python
print("Median order amount:", df['total_items'].median())
print("Median order amount:", df['total_items'].mean())
print("Mode order amount:", df['total_items'].mode().iat[0])
```
